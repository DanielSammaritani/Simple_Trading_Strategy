```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import yfinance as yf
plt.style.use("seaborn-v0_8")
tickers = [
    "IXN",
    "IXJ",
    "IXG",
    "IXC",
    "RXI",
    "KXI",
    "EXI",
    "MXI",
    "IXP",
    "JXI",
    "REET"
]

data = yf.download(tickers ,start="2020-01-01")
print(data)
```

```python
plt.figure(figsize=(12,6))
plt.plot(data["Close"])
plt.legend(tickers)
data.dropna(inplace=True)
plt.show()
```

```python
Price = data["Close"]
def cum_returns(x):
    ret = np.exp(x.sum())-1
    return ret
def Sharpe(x):
    annualized_growth = np.exp(x.mean()*252)-1
    risk_free_rate_ann = np.exp(0.048)-1
    annualized_volatility = np.sqrt(252)*x.std()
    Sharpe = (annualized_growth-risk_free_rate_ann)/annualized_volatility
    return Sharpe
def max_drawdown(x):
    cum_log_ret = x.cumsum()
    equity_line = np.exp(cum_log_ret)
    d_max = equity_line.cummax()
    d = (d_max-equity_line)/d_max
    return d.max()



```

```python
print(Price)
```

```python
#Strategy
K=2
days = 20
SMA = Price.rolling(days).mean()
std = Price.rolling(days).std()
Hb = SMA +(2*std)
Lb = SMA -(2*std)
print(Hb)
```

```python
signals = pd.DataFrame(np.select([Price > Hb, Price < Lb], [-1,+1], default = 0), index = Price.index, columns = Price.columns)
print(signals)
```

```python
log_ret = np.log(data["Close"]/data["Close"].shift(1)).dropna()
strat = log_ret* signals.shift(1)
print(strat)
```

```python
metrics = {
    "Cum_returns" : cum_returns(log_ret),
    "Sharpe" : Sharpe(log_ret),
    "Max Drawdown" : max_drawdown(log_ret)

}
dashboard_bh = pd.DataFrame(metrics)
print(dashboard_bh)
```

```python
metrics1 = {
    "Cum_returns" : cum_returns(strat),
    "Sharpe" : Sharpe(strat),
    "Max Drawdown" : max_drawdown(strat)
}
dashboard_strat = pd.DataFrame(metrics1)
print(dashboard_strat)
```
